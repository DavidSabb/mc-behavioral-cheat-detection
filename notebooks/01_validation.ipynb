```python
import pandas as pd
import numpy as np

df = pd.read_csv("../server/plugins/Collector/movement.csv")
s = df[df.session_id == df.session_id.iloc[-1]].copy()

print("rows:", len(s))
print("duration s:", (s.ts.max() - s.ts.min()) / 1000)
print("rows/sec:", len(s) / ((s.ts.max() - s.ts.min()) / 1000))

s["dt"] = s.ts.diff()
print("\ninter-row gap (ms):")
print(s.dt.describe())

s["dx"] = s.x.diff(); s["dz"] = s.z.diff()
s["speed"] = np.hypot(s.dx, s.dz)
print("\nblocks per tick:")
print(s.speed[s.speed > 0].describe())



print(s.dt.sort_values(ascending=False).head(10))
```

```python
s["dyaw"] = s.yaw.diff().apply(lambda a: (a + 180) % 360 - 180)
print(s.dyaw.describe())
print("nonzero fraction:", (s.dyaw.abs() > 1e-6).mean())
```

```python
k = pd.read_csv("../server/plugins/Collector/clicks.csv")
iv = k[k.session_id == k.session_id.iloc[-1]].ts.diff().dropna()
print("CPS:", 1000 / iv.mean())
print("interval std:", iv.std())
print("min interval:", iv.min())
```
